```python
# Hollow Square Pattern
n = int(input("Enter the side length of the square: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()  
```

```python
# Hollow Rectangle Pattern
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

for i in range(rows):
    for j in range(cols):
        if i == 0 or i == rows-1 or j == 0 or j == cols-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()  
```

```python
# Hollow Right-Angled Triangle
n = int(input("Enter the height of the triangle: "))

for i in range(1, n+1):  # rows
    for j in range(1, i+1):  # columns
        if j == 1 or j == i or i == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()  
```

```python
# Hollow Inverted Right-Angled Triangle
n = int(input("Enter the height of the triangle: "))

for i in range(n, 0, -1):
    for j in range(1, i+1):  # columns
        if j == 1 or j == i or i == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print() 
```

```python
# Hollow Pyramid Pattern
n = int(input("Enter the height of the pyramid: "))

for i in range(1, n+1):  # rows
    for j in range(n-i):
        print(" ", end=" ")
    for j in range(1, 2*i):
        if j == 1 or j == 2*i-1 or i == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()  
```

```python
# Hollow Inverted Pyramid Pattern
n = int(input("Enter the height of the inverted pyramid: "))

for i in range(n, 0, -1):  # rows from n down to 1
    for j in range(n - i):
        print(" ", end=" ")
    for j in range(1, 2*i):
        if j == 1 or j == 2*i-1 or i == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Hollow Cross (Plus Sign) Pattern
n = int(input("Enter the size of the grid (odd number): "))

for i in range(n):  # rows
    for j in range(n):  # columns
        if i == n//2 or j == n//2:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print() 
```
