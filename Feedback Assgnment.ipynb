```python
# Function to write feedback
def write_feedback():
    positive_keywords = ["good", "excellent", "amazing", "great", "fantastic", "love"]
    negative_keywords = ["bad", "poor", "terrible", "hate", "awful", "worst"]

    try:
        feedback = input("Enter your feedback: ").strip().lower()

        # Categorize feedback
        if any(word in feedback for word in positive_keywords):
            filename = "good_feedback.txt"
        elif any(word in feedback for word in negative_keywords):
            filename = "bad_feedback.txt"
        else:
            filename = "neutral_feedback.txt"

        # Write feedback to file
        with open(filename, "a") as file:
            file.write(feedback + "\n")

    except Exception as e:
        print(f"Error while writing feedback: {e}")
    else:
        print("Thank you for your feedback!")
    finally:
        print("Feedback submission process completed.")
```
